```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def show_result(self):
        print(f"{self.name}: {self.score}점")

    def is_passed(self):
        return self.score >= 60
```

```python
student1 = Student("홍길동", 45)

print(f"{student1.name}님의 점수는 {student1.score}입니다.")

if student1.is_passed() == True:
    # 위에 def is_passed(self):여기에 parameter가 self 제외하고는 없으므로, ()

    print(f"{student1.name}님은 이번 시험에서 당당히 합격하셨습니다.")
else :
    print(f"{student1.name}님은 이번 시험에서 아쉽지만 탈락하셨습니다.")
```

```python
class Character:
    def __init__ (self, name, level):
        self.name = name
        self.level = level
    def show_info(self):
        print(f"[{self.name}] 현재 레벨 : {self.level}")

    def is_expert(self):
        return self.level >= 10
        
```

```python
char1 = Character("전사", 12)
char2 = Character("초보자", 5)

char1.show_info()
print(char1.is_expert())

char2.show_info()
print(char2.is_expert())
```
