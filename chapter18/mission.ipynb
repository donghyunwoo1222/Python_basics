```python
# MISSION 01 — 통과 여부 메서드 만들기
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def show_result(self):
        print(f"{self.name}의 점수는 {self.score}입니다.")
    def is_passed(self):
        return self.score >= 60

student1 = Student("민수", 85)
student2 = Student("서준", 55) 

student1.show_result()
student1.is_passed()
student2.show_result()
student2.is_passed()
```

```python
# MISSION 02 — 상품 객체 리스트 처리하기

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def show_info(self):
        print(f"{self.name} : {self.price}원")

product1 = Product("노트북", 1200000)
product2 = Product("무선 마우스", 35000)
product3 = Product("기계식 키보드", 89000)
product4 = Product("모니터", 250000)

products = [product1, product2, product3, product4]

for product in products:
    product.show_info()
```

```python
# 종합 실습 — 상품 재고 객체 만들기

class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def sell(self, quantity):
        if quantity <= self.stock:
            self.stock -= quantity
            return True
        return False

    def show_info(self):
        print(f"{self.name} : {self.price}원 / 재고 {self.stock}개")

product = Product("키보드", 45000, 10)
product.show_info()
product.sell(4)
product.show_info()
```

```python
# MISSION 03 — 재고 부족 상황 추가하기

class Product:
    def __init__(self,name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def sell(self, request):
        if request <= self.stock:
            self.stock-=request
            return True
        return False
    
    def show_info(self):
        print(f"{self.name}: {self.price}원 / 재고 {self.stock}개")


product = Product("마우스", 50000, 15)
product.show_info()
product.sell(16)
product.show_info()
```
