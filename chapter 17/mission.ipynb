```python
# SELF MISSION — 안전한 점수 입력 프로그램

name = input("이름을 입력하세요 : ")
while True:
    try:
        score = int(input("점수를 입력하세요 : "))
    except ValueError:
        print("숫자를 입력하세요")
        continue
    if 0<=score<=100:
        break
    else:
        print("0과 100 사이의 숫자를 입력하세요")

print(f"{name} :  {score}")




```
