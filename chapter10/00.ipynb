# for와 while 비교

- for   → 반복 횟수나 반복할 값이 비교적 분명할 때
- while → 특정 조건이 True인 동안 계속할 때

```python
count = 1

while count <= 3:

    print(f"현재 count: {count}")

    count += 1

print("종료")
```

```python
# break는 반복 전체를 끝낸다.
for number in range(1, 6):

    if number == 3:

        break

    print(number)
#여기서는 3이 되자마자 break 되므로, 3도 출력되지 않음
# 1 → 출력
# 2 → 출력
# 3 → break
# 4 → 실행 안 함
# 5 → 실행 안 함
```

```python
# continue는 이번 반복만 건너뛴다.
for i in range(1,6):
    if i == 3:
        continue
    print(i)
```

# break와 비교
 
break    → 반복 전체 종료
continue → 현재 반복만 건너뛰고 다음 반복 계속

```python
# 반복문 안에 반복문 
for row in range(1,3):
    for column in range(1,4):
        print(f"row={row}, column={column}")
```

row = 1

→ column 1, 2, 3 전체 반복

row = 2

→ column 1, 2, 3 전체 반복
