```python
def greet():
    print("안녕하세요!")
    print("파이썬 공부를 시작합니다.")

# greet() 이라는 함수를 만든다. 그 안에 2,3번째 줄이 들어가 있는 것
greet()
greet()
# greet()이라는 함수를 불러온다. 여기서는 greet()함수가 2,3번째 줄을 실행하기 때문에 아래처럼 결과가 나온다. 
```

```python

def greet(name):
    print(f"안녕하세요! {name}")
    print("파이썬 공부를 시작합니다.")
greet("홍길동")
greet("우동현")
```

```python

def greet(name, subject):
    print(f"안녕하세요! {name}님")
    print(f"{subject} 공부를 시작합니다.")
greet("홍길동","파이썬") #파이썬
greet("우동현","SQL") #SQL



def greet(name, subject):
    print("안녕하세요!", name,"님")
    print(subject, "공부를 시작합니다.")
greet("홍길동","파이썬") #파이썬
greet("우동현","SQL") #SQL
```

```python
# 두개의 숫자를 입력 받아서 더한 값을 반환하는 함수 

#함수 이름 : first_sum

def first_sum(first_n, second_n):
    return first_n + second_n

num1 = int(input("첫번재 숫자 : "))
num2 = int(input("첫번재 숫자 : "))

result = first_sum(num1, num2)

print(f"두 수의 합은 {result}입니다.")
```

# print()와 return()은 역할이 다르다. 

print는 그냥 사용자에게 보여주는거고, return은 값을 함수 밖으로 전달해서 저장/계산에 재사용한다. 

```python
# 두 수(1~9)를 입력 받아서 사칙연산 결과를 반환하는 calc 라는 함수를 만드시오.
# 4개 결과값을 한 번에 반환(Tuple)
# 사용자가 입력한 값이 숫자가 아닐 경우 오류메시지 출력하고 프로그램 종료 

def calc(a,b):
    calc_sum = a+b
    calc_difference = a-b
    calc_multiple = a*b
    calc_division = a/b
    return calc_sum, calc_difference, calc_multiple, calc_division

num1 = input("첫번재 숫자 : ")
num2 = input("두번재 숫자 : ")

lenth = len(num1) + len(num2)

if (lenth ==2 and ord(num1) >= 49 and ord(num1) <= 57) and (lenth ==2 and ord(num2)>= 49 and ord(num2)<= 57):
    # 제대로 둘 다 숫자로 변환 가능
    
    result = calc(int(num1),int(num2))
    print(type(result))
    print(result) 
else:
    print("입력한 값(숫자)을 확인하세요!")


```
