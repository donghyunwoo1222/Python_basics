```python
# MISSION 01 — 합계와 평균 함께 반환하기

def calculate_summary(a,b):
    total = a+b
    avg = (a+b)/2
    return total, avg

total, average = calculate_summary(80, 100)
print(f"total = {total}")
print(f"average = {average}")
```

```python
# MISSION 02 — 리스트 평균 함수 만들기

def calculate_average(scores):
    total = 0
    
    for score in scores:
        total += score
    average = total / len(scores)
    return average

scores = [85, 92, 78]
avg = calculate_average(scores)
print(avg)
```

```python
# 종합 실습 — 학생 점수 보고서 함수 만들기

def calculate_average(scores):
    total = 0
    for score in scores:
        total+=score
    avg = total / len(scores)
    return avg

def get_grade(score):
    if score >= 95:
        return("A")
    if score >= 90:
        return("B")
    else:
        return("C")

def show_report(name, average, grade):
    print(f"{name}님의 평균은 {average}이고, 등급은 {grade}입니다.")


name = "민수"
scores = [85, 90, 95]

average = calculate_average(scores)
grade = get_grade(average)
show_report(name, average, grade)

```
