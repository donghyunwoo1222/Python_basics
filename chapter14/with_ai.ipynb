2. 특정 문자 개수 세기 (count_character)

- 요구사항: 문자열과 찾고 싶은 특정 문자(알파벳)를 매개변수로 받아, 문자열 안에서 그 문자가 총 몇 번 등장하는지 세어서 숫자로 반환하는 함수를 작성하세요.

- 매개변수: text, target (예: text="banana", target="a")

- 반환값: 정수형 숫자 (예: 3)

- 힌트: for 반복문으로 문자열을 한 글자씩 돌면서 조건문(if)을 통해 일치하는지 확인해 보세요.

```python
#1. 
def get_score_stats(scores):

    maximum = max(scores)
    minimum= min(scores)
    average = sum(scores)/len(scores)
    return{"max": maximum,"min": minimum,"average": average}


scores = [80, 90, 70, 100, 85]
get_score_stats(scores)
```
