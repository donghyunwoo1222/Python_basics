```python
# Part1. 리스트와 반복문을 이용한 데이터 처리

products = ["사과", "바나나", "체리", "포도", "오렌지", "체리"]

count = 0
for product in products:
    # 조건에 맞는 데이터 찾기(체리)
    if product == "체리":
        count += 1
    
print(f"{product} : {count}개")
# 조건을 만족하는 데이터 개수 계산
        

# 여기서 의문점. 이런식으로 코드를 세우면 if product =="바나나" 이런식으로 고쳐도 계속 체리가 출력됨
# 생각을 해보니 products를 하나씩 반복할때 리스트의 마지막 값이 체리이기 때문에 체리가 출력되는 것
# 따라서 실제로 출력값은 맞지만 제대로 된 코드가 아니다. 

# AI의 도움을 받음
# "내가 찾는 product를 나타내는 다른 변수를 지정해두고, product==변수 이렇게 코드를 세우면 해결됨"

#새로 세운 코드
products = ["사과", "바나나", "체리", "포도", "오렌지", "체리"]

count = 0
target = "사과"
for product in products:
    # 조건에 맞는 데이터 찾기(바나나)
    if product == target:
        count += 1
    
print(f"{target} : {count}개")
```

```python
# Part 2. 딕셔너리를 이용한 단어 빈도 분석
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

blank_dict = {}

for word in sentence.split():
    if word in blank_dict:
        blank_dict[word]+=1
    else:
        blank_dict[word]=1
    
for word, count in blank_dict.items():
    #전체 단어와 등장 횟수를 출력합니다.
    print(f"{word} : {count}회")

# 2회 이상 등장한 단어만 별도로 출력합니다.
for word, count in blank_dict.items():
    if count>=2:
        print((f"{word} : {count}회"))


# 빈 딕셔너리를 만들어 split으로 sentence를 단어로 나누어 반복문에 넣는건 스스로 했다.
# 그 이후에 막혀서 AI를 통해 items메서드를 사용해야한다는 힌트를 받아 해결함


```

```python
# Part 3. 함수로 반복되는 코드 정리하기

def count_words(sentence):
    blank_dict = {}
    for word in sentence.split():
        if word in blank_dict:
            blank_dict[word]+=1
        else:
            blank_dict[word]=1
    return blank_dict
    # 함수내부에서 단어 빈도 계산 후 return

sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

# count_words 함수를 parameter sentence 사용해서 함수 호출
result = count_words(sentence)
print(result) 

# 위에서 사용했던 형식 그대로 함수로 만들어서 return 후 함수 밖에서 호출해서 출력하는 것까지 실행함

# 2회 이상 등장한 단어만 출력하는 기능도 별도 함수로 분리

def print_words(result):
    for word, count in result.items():
        if count>=2:
            print(f"{word} : {count}회")

print_words(result)

# 지금까지 return으로 함수 밖에서 출력하는 것만 해봐서 print를 활용해본 적이 거의 없어서 해결이 안됐다.
# ai를 통해 힌트를 얻어서 단순하게 조건에 맞는 것을 바로 출력하기에는 print()가 간단하다는 것을 배우고 해결함
# return을 통해 함수 안에 빈 딕셔너리를 만들고, 딕셔너리를 채운 후 return해서 함수 밖에서 출력하는 방법도 있지만
# print를 통해 함수 내에서 출력하는게 이 경우엔 더 간단한 것 같다. 앞으로 함수는 공부를 더 해야할 것 같다.
     
```
