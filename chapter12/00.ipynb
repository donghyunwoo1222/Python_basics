# set(집합)
- 고유한 값만 필요하다면 중복을 직접 찾기보다는 집합을 사용한다.
- 집합에서는 같은 값이 여러 번 있어도 하나만 남는다.
- 집합의 출력 순서는 인덱스 순서처럼 생각하면 안됨. 순서가 없음 -> 인덱스 사용불가
* 빈 집합은 {}가 아니라 set(). {}은 빈 딕셔너리

```python
# 집합에 값 추가하고 삭제하기
skills = {"Python", "SQL"}
skills.add("AI")
print(skills)
skills.remove("SQL")
print(skills)

# 없는 값을 remove 하려고 하면 KeyError가 발생한다. 
# 없는 값이라도 오류없이 넘어가고 싶으면 discard()를 사용한다.
skills.discard()
```
