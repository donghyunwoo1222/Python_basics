```python
# 종합 실습 — 두 팀의 기술 목록 비교하기 

team_a_list = ["Python", "SQL", "AI", "Python"]
team_b_list = ["Python", "Java", "AI", "Java"]

team_a = set(team_a_list)
team_b = set(team_b_list)
print("A팀 고유 기술 :", team_a)
print("B팀 고유 기술 :", team_b)
print("공통 기술 :",team_a&team_b)
print("전체 기술 :",team_a|team_b)
print("A팀만 :",team_a-team_b)
print("B팀만 :",team_b&team_a)
```
