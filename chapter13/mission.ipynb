```python
# 종합실습 

students = [
    {"name": "민수", "score": 85, "city": "서울"},
    {"name": "지영", "score": 92, "city": "부산"},
    {"name": "서준", "score": 78, "city": "서울"}
]
#STEP 1 — 따라해 보기
#모든 학생의 이름과 점수를 출력합니다.
for student in students: 
    print(student["name"],student["score"])

#STEP 2 — 바꿔 보기
#80점 이상 학생만 출력합니다.
for student in students:
    if student["score"] >= 80:
        print(student)

#STEP 3 — 조건 추가
#서울에 사는 학생만 출력합니다.
for student in students:
    if student["city"] == "서울":
        print(student)

#STEP 4 — 두 조건 연결
#서울에 살면서 80점 이상인 학생을 출력합니다.
for student in students:
    if student["score"] >= 80 and student["city"] == "서울":
        print(student)
```
