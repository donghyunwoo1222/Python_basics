```python
# MISSION 02 — 80점 이상 학생만 출력하기

students = [
    {"name": "민수", "score": 85},
    {"name": "지영", "score": 92},
    {"name": "서준", "score": 78}
]

for student in students:
    if student["score"] >= 80:
        print(student["name"],student["score"])
```

```python
# 종합 실습 — 학생 성적 정보 관리하기

students = [
    {"name": "민수", "score": 85, "city": "서울"},
    {"name": "지영", "score": 92, "city": "부산"},
    {"name": "서준", "score": 78, "city": "서울"}
]

for student in students:
    if student["score"] >= 80 and student["city"] == "서울":
        print(student["name"],student["score"], student["city"])

print(type(student["score"]))
print(student.get("city"))
```
